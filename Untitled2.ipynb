```python
from google.colab import files
uploaded = files.upload()

dataset_file = list(uploaded.keys())[0]
print("Dataset uploaded:", dataset_file)
```

```python
import json

texts = []

with open(dataset_file, "r", encoding="utf-8") as f:
    try:
        data = json.load(f)
        if isinstance(data, list):
            for item in data:
                texts.append(str(item))
    except:
        f.seek(0)
        for line in f:
            texts.append(str(json.loads(line)))

print("Dataset loaded with", len(texts), "resumes")
```

```python
uploaded_resume = files.upload()
test_file = list(uploaded_resume.keys())[0]

print("Test file:", test_file)
```

```python
from docx import Document
import PyPDF2

def extract_text(file_name):
    text = ""

    if file_name.endswith(".docx"):
        doc = Document(file_name)
        for para in doc.paragraphs:
            text += para.text + " "

    elif file_name.endswith(".pdf"):
        pdf = PyPDF2.PdfReader(file_name)
        for page in pdf.pages:
            text += page.extract_text() or ""

    return text
```

```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification")
```

```python
def predict_resume(file_name):
    text = extract_text(file_name)

    from transformers import pipeline
    classifier = pipeline("zero-shot-classification")

    labels = ["Genuine resume", "Fake or exaggerated resume"]
    result = classifier(text, candidate_labels=labels)

    print("AI Result:", result)

    fake_score = result['scores'][result['labels'].index("Fake or exaggerated resume")]

    print("Fake Score:", fake_score)

    # FINAL DECISION (THIS WILL WORK)

    # If AI is confident fake
    if fake_score > 0.5:
        return "Fake Resume ❌"

    # Detect unrealistic skill overload (automatic)
    if text.count("Machine Learning") + text.count("Deep Learning") + text.count("AI") > 5:
        return "Fake Resume ❌"

    return "Genuine Resume ✅"
```

```python
result = predict_resume(test_file)
print("Final Result:", result)
```
